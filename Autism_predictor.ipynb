```python
# Calculate outliers

Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3-Q1

lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['age'] > upper_bound) | (df['age'] < lower_bound)]
```

```python
X = encoded_df.drop('Class/ASD' , axis=1)
y = encoded_df['Class/ASD']
```

```python
# train test split
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size=0.33, random_state=42)
```

```python
# train base madel of classification problem
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
```

```python
lr.fit(X_train , y_train)
```

```python
lr_test = lr.predict(X_test)
lr_train = lr.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score , precision_score , recall_score , f1_score , confusion_matrix
```

```python
print("Logistic Testing", accuracy_score(y_test , lr_test))
print("Logistic Training",accuracy_score(y_train , lr_train))
print("Lgistic Precision Test" , precision_score(y_test , lr_test))
print("Lgistic Precision train" , precision_score(y_train , lr_train))
print("Lgistic recall_score Test" , recall_score(y_test , lr_test))
print("Lgistic recall_score train" , recall_score(y_train , lr_train))
print("Logistic f1 score test" , f1_score(y_test , lr_test))

print(confusion_matrix(y_test , lr_test))
```

```python
# Now check another model
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train , y_train)
```

```python
knn_test = knn.predict(X_test)
knn_train = knn.predict(X_train)
```

```python
print("knn Testing", accuracy_score(y_test , knn_test))
print("knn Training",accuracy_score(y_train , knn_train))
print("knn Precision Test" , precision_score(y_test , knn_test))
print("knn Precision train" , precision_score(y_train , knn_train))
print("knn recall_score Test" , recall_score(y_test , knn_test))
print("knn recall_score train" , recall_score(y_train , knn_train))
print("knn f1 score test" , f1_score(y_test , knn_test))

print("KNN Confusion matrix " , confusion_matrix(y_test , knn_test))
```

## Finding the Best K for KNN

```python
results = []
for k in range(1, 11):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)

    knn_test = knn.predict(X_test)
    knn_train = knn.predict(X_train)

    results.append({
        'k_value': k,
        'accuracy_test': accuracy_score(y_test, knn_test),
        'accuracy_train': accuracy_score(y_train, knn_train),
        'precision_test': precision_score(y_test, knn_test),
        'precision_train': precision_score(y_train, knn_train),
        'recall_test': recall_score(y_test, knn_test),
        'recall_train': recall_score(y_train, knn_train),
        'f1_test': f1_score(y_test, knn_test)
    })

results_df = pd.DataFrame(results)
display(results_df)
```
